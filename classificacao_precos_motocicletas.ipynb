```python
X = df.drop(columns=['selling_price'])
y = df['selling_price']

print("Formato das variáveis preditoras:", X.shape)
print("Formato da variável alvo:", y.shape)
```

```python
X_treino, X_temp, y_treino, y_temp = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42
)
```

```python
X_validacao, X_teste, y_validacao, y_teste = train_test_split(
    X_temp,
    y_temp,
    test_size=0.50,
    random_state=42
)
```

```python
print("Treinamento:", X_treino.shape)
print("Validação:", X_validacao.shape)
print("Teste:", X_teste.shape)
```

```python
limite_1 = y_treino.quantile(1/3)
limite_2 = y_treino.quantile(2/3)

print("Limite econômico/intermediário:", limite_1)
print("Limite intermediário/premium:", limite_2)
```

```python
def classificar_preco(precos):
    return pd.cut(
        precos,
        bins=[-np.inf, limite_1, limite_2, np.inf],
        labels=['Economica', 'Intermediaria', 'Premium'],
        include_lowest=True
    )
```

```python
y_treino = classificar_preco(y_treino)
y_validacao = classificar_preco(y_validacao)
y_teste = classificar_preco(y_teste)
```

```python
print("Treinamento:")
print(y_treino.value_counts())

print("\nValidação:")
print(y_validacao.value_counts())

print("\nTeste:")
print(y_teste.value_counts())
```

```python
colunas_numericas = [
    'year',
    'km_driven',
    'ex_showroom_price'
]

colunas_categoricas = [
    'name',
    'seller_type',
    'owner'
]

print("Variáveis numéricas:", colunas_numericas)
print("Variáveis categóricas:", colunas_categoricas)
```

```python
preprocessador_numerico = Pipeline(steps=[
    ('imputador', SimpleImputer(strategy='median'))
])

preprocessador_categorico = Pipeline(steps=[
    ('imputador', SimpleImputer(strategy='most_frequent')),
    ('codificador', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
preprocessador = ColumnTransformer(transformers=[
    ('numericas', preprocessador_numerico, colunas_numericas),
    ('categoricas', preprocessador_categorico, colunas_categoricas)
])
```

```python
modelo_arvore = Pipeline(steps=[
    ('preprocessamento', preprocessador),
    ('classificador', DecisionTreeClassifier(
        random_state=42,
        max_depth=10
    ))
])

modelo_floresta = Pipeline(steps=[
    ('preprocessamento', preprocessador),
    ('classificador', RandomForestClassifier(
        n_estimators=200,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
modelo_arvore.fit(X_treino, y_treino)

modelo_floresta.fit(X_treino, y_treino)

print("Modelos treinados com sucesso!")
```

```python
previsao_arvore = modelo_arvore.predict(X_validacao)

previsao_floresta = modelo_floresta.predict(X_validacao)

acuracia_arvore = accuracy_score(
    y_validacao,
    previsao_arvore
)

acuracia_floresta = accuracy_score(
    y_validacao,
    previsao_floresta
)

print(f"Acurácia - Árvore de Decisão: {acuracia_arvore:.4f}")
print(f"Acurácia - Random Forest: {acuracia_floresta:.4f}")
```

```python
comparacao = pd.DataFrame({
    'Modelo': ['Árvore de Decisão', 'Random Forest'],
    'Acurácia': [acuracia_arvore, acuracia_floresta]
})

display(
    comparacao.sort_values(
        by='Acurácia',
        ascending=False
    )
)
```

```python
### Comparação dos modelos

Foram treinados dois algoritmos de classificação:
Árvore de Decisão e Random Forest.

Na avaliação utilizando os dados de validação,
a Árvore de Decisão apresentou acurácia de 73,42%,
enquanto o Random Forest alcançou 74,68%.

O Random Forest apresentou o melhor resultado,
com uma diferença de aproximadamente 1,27 ponto
percentual em relação à Árvore de Decisão.

Dessa forma, o Random Forest foi selecionado
para a avaliação final.
```

```python
previsao_teste = modelo_floresta.predict(X_teste)

acuracia_final = accuracy_score(
    y_teste,
    previsao_teste
)

print(f"Acurácia final: {acuracia_final:.4f}")
print(f"Acurácia final: {acuracia_final * 100:.2f}%")
```

```python
print(classification_report(
    y_teste,
    previsao_teste,
    zero_division=0
))
```

```python
matriz = confusion_matrix(
    y_teste,
    previsao_teste,
    labels=['Economica', 'Intermediaria', 'Premium']
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=matriz,
    display_labels=['Econômica', 'Intermediária', 'Premium']
)

fig, ax = plt.subplots(figsize=(8, 6))

disp.plot(
    cmap='Blues',
    values_format='d',
    ax=ax
)

plt.title('Matriz de Confusão - Random Forest')
plt.show()
```

```python
amostra = X_teste.iloc[[0]]

classe_real = y_teste.iloc[0]
classe_prevista = modelo_floresta.predict(amostra)[0]

print("Dados da motocicleta:")
display(amostra)

print("Categoria real:", classe_real)
print("Categoria prevista:", classe_prevista)
```

```python
### Avaliação final do modelo

O modelo Random Forest foi selecionado após apresentar
a maior acurácia na etapa de validação, alcançando 74,68%.

Na avaliação final, utilizando os dados de teste,
o modelo obteve acurácia de 78,62%.

O relatório de classificação apresentou F1-score
ponderado de 79%, demonstrando um desempenho
satisfatório na identificação das três categorias.

A categoria Premium apresentou o melhor resultado,
com F1-score de 86%, enquanto a categoria
Intermediária apresentou maior dificuldade de
classificação.

Conclui-se que o modelo conseguiu identificar
as faixas de preço das motocicletas com desempenho
próximo de 79% de acerto no conjunto de teste.
```
